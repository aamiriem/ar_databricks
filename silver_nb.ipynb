```python
%sql
select join_date, date_format(now(), 'yyyy-MM-dd') as mday, concat( timestampdiff(year, join_date, now()),' ','Year ',
timestampdiff(month, join_date, now())%12,' ', 'Month ',
case
when day(join_date)>day(now()) then day(join_date)- day(now())
else day(now())-day(join_date)
end,' ' 'Days') as AGE

from employee
```

```python
%sql
select from_utc_timestamp(now(), 'Asia/Kolkata'),date_format(now(),'dd-MM-yyyy'),date_format(date_add(day,1050,now()),'dd-MM-yyyy') as after6,
case
when dayofweek(date_add(day,1050,now()))=1 then 'Sunday'
when dayofweek(date_add(day,1050,now()))=2 then 'Monday'
else 'Others'
end as Day
```
